```python
import random
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow # for image display
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pylab as plt
```

```python
# Membuat list untuk menyimpan url dari beberapa image
urls = ["https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
        "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg",
        "https://asset.kompas.com/crops/eF0UAAFJU3nALkvwWqPX_MR74Ik=/0x0:1200x800/1200x800/data/photo/2023/08/03/64cb16c44b560.jpg"]

# baca dan tampilkan image
# loop pada tiap url image, beberapa image dapat disimpan pada list
for url in urls:
    image = io.imread(url)                        #read image
    image = cv.resize(image, (0,0), fx=0.5, fy=0.5) #resize image to half size
    image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB) #convert color to RGB
    final_frame = cv.hconcat((image, image_2))     #concatenate image
    cv2_imshow(final_frame)                        #show image
    print('\n')
```

```python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image: tinggi x lebar = ",tinggi," x ",lebar)
cv2_imshow(image_2)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

#membuat garis horizontal ditengah image
for y in range (lebar):
    image_3[int((tinggi)/2),y] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

2.Buat garis vertikal dan garis menyilang diagonal pada image keluaran

VERTIKAL

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)


# Membuat garis vertikal dengan loop (menggunakan pembagian bulat //)
for x in range(tinggi):
    image_3[x, lebar // 2] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

SILANG

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# Garis Diagonal (Menyilang)
y_coords = np.linspace(0, tinggi - 1, num=lebar).astype(int)
x_coords = np.arange(lebar)

# Diagonal dari kiri-atas ke kanan-bawah
image_3[y_coords, x_coords] = [255, 255, 255]

# Diagonal dari kiri-bawah ke kanan-atas
image_3[tinggi - 1 - y_coords, x_coords] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

3.Buat garis horisontal berwarna putih dibagian tengah gambar dengan panjang tertentu


```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# Ambil tinggi dan lebar gambar dari shape
tinggi, lebar = image.shape[:2]

# Hitung titik tengah vertikal (Y)
y_tengah = tinggi // 2

# Menentukan panjang garis yang diinginkan (100 piksel)
panjang_garis = 100

# Hitung koordinat awal (X1) dan akhir (X2) agar posisi garis presisi di tengah
start_x = max(0, (lebar - panjang_garis) // 2)
end_x = min(lebar, start_x + panjang_garis)

# Gambar garis putih pada rentang koordinat X tersebut
image_3[y_tengah, start_x:end_x] = [255, 255, 255]

# Gabungkan dan tampilkan gambar
final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

4.Buat kotak menggunakan kumpulan pixel warna putih di sembarang tempat dalam gambar

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

#Ambil tinggi dan lebar dari variabel image
tinggi, lebar = image.shape[:2]

image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# Menentukan ukuran kotak (lebar 80px, tinggi 50px)
kotak_lebar = 80
kotak_tinggi = 50

# Menentukan koordinat titik awal (kiri-atas kotak) secara acak
start_y = random.randint(0, tinggi - kotak_tinggi)
start_x = random.randint(0, lebar - kotak_lebar)

# Hitung koordinat titik akhir
end_y = start_y + kotak_tinggi
end_x = start_x + kotak_lebar

# Warnai area kotak tersebut dengan piksel warna putih
image_3[start_y:end_y, start_x:end_x] = [255, 255, 255]

# Tampilkan hasil
final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```
