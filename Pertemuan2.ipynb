Selain itu, program harus bersifat adaptif, dimana setiap ukuran, posisi, atau ketebalan harus dihitung dari ukuran citra masukan, bukan ditulis sebagai angka tetap pada kode program.

Studi Kasus 1: Panitia sebuah kegiatan perlu membubuhkan label identitas pada ratusan foto dokumentasi. Foto berasal dari beberapa kamera dan ponsel, sehingga ukurannya bermacam-macam. Label harus terlihat seimbang pada semua foto tanpa disetel satu per satu. Fungsi yang sama dikenakan pada dua citra berbeda ukuran. Tinggi bar dan ukuran huruf ikut menyesuaikan, sehingga proporsinya tetap sama.

jawab:


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def tambah_label_identitas(img):
    # 1. Ambil dimensi citra (tinggi, lebar, kanal)
    h, w = img.shape[:2]

    # Buat salinan agar citra asli tidak tertimpa
    img_out = img.copy()

    # 2. Hitung dimensi bar secara adaptif (misal: tinggi bar 10% dari tinggi citra)
    bar_height = int(h * 0.10)

    # 3. Buat bar transparan/solid di bagian bawah citra
    # Koordinat pt1 (kiri-atas bar) dan pt2 (kanan-bawah bar)
    pt1 = (0, h - bar_height)
    pt2 = (w, h)
    cv2.rectangle(img_out, pt1, pt2, color=(0, 0, 0), thickness=-1)

    # 4. Hitung skala font, ketebalan, dan posisi teks secara adaptif
    # Skala font disesuaikan dengan lebar citra
    font_scale = w / 800.0  # Basis proporsi untuk skala 1.0 pada lebar 800px
    font_thickness = max(1, int(w / 400))

    teks = "DOKUMENTASI KEGIATAN - 2026"
    font = cv2.FONT_HERSHEY_SIMPLEX

    # Hitung ukuran teks dalam piksel untuk pemosisian presisi (centered)
    (text_w, text_h), baseline = cv2.getTextSize(teks, font, font_scale, font_thickness)

    # Posisi teks di tengah bar bawah
    text_x = int((w - text_w) / 2)
    text_y = int(h - (bar_height - text_h) / 2)

    # 5. Tuliskan teks label pada bar
    cv2.putText(img_out, teks, (text_x, text_y), font, font_scale, (255, 255, 255), font_thickness)

    return img_out

# === PENGUJIAN PADA DUA CITRA BERBEDA UKURAN ===

# 1. Baca citra asli (misal: Citra A)
img_a = cv2.imread('/content/drive/MyDrive/PCVK/729ad62c-aa04-4126-9758-7dc8d2a17e7d.jpg')
img_a = cv2.cvtColor(img_a, cv2.COLOR_BGR2RGB)

# 2. Buat Citra B dengan ukuran berbeda (misal: 50% dari ukuran Citra A menggunakan cv2.resize)
h_orig, w_orig = img_a.shape[:2]
img_b = cv2.resize(img_a, (int(w_orig * 0.5), int(h_orig * 0.5)))

# 3. Jalankan fungsi adaptif pada kedua citra
hasil_a = tambah_label_identitas(img_a)
hasil_b = tambah_label_identitas(img_b)

# 4. Tampilkan kedua hasil secara berdampingan
fig, axes = plt.subplots(1, 2, figsize=(12, 6))

axes[0].imshow(hasil_a)
axes[0].set_title(f'Citra A (Ukuran Besar: {img_a.shape[1]}x{img_a.shape[0]})')
axes[0].axis('off')

axes[1].imshow(hasil_b)
axes[1].set_title(f'Citra B (Ukuran Kecil: {img_b.shape[1]}x{img_b.shape[0]})')
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

Studi Kasus 2: Marketplace seperti Tokopedia dan Shopee menampilkan foto produk dalam bingkai persegi 1:1. Foto penjual yang berukuran potrait atau lanskap akan terpotong atau terlihat tidak proporsional bila diunggah begitu saja. Bangun program yang mengubah foto berukuran apa pun menjadi thumbnail persegi tanpa mengubah rasio aslinya, dengan sisa ruang diisi kanvas berwarna. Dalam hal ini, rasio asli tetap terjaga, citra terpasang tepat di tengah kanvas, tambahkan logo nama toko berukuran proporsional pada gambar.

jawab:

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def buat_thumbnail_persegi(img, nama_toko="STORE OFFICIAL"):
    # 1. Ambil dimensi gambar asli
    h, w = img.shape[:2]

    # 2. Tentukan ukuran kanvas persegi berdasarkan dimensi terpanjang (1:1)
    max_dim = max(h, w)

    # 3. Buat kanvas kosong berwarna netral (misal: abu-abu terang / light gray)
    # Kanvas berukuran (max_dim, max_dim, 3)
    kanvas = np.full((max_dim, max_dim, 3), 240, dtype=np.uint8)

    # 4. Hitung posisi offset agar gambar berada tepat di tengah (center placement)
    x_offset = (max_dim - w) // 2
    y_offset = (max_dim - h) // 2

    # 5. Tempelkan gambar asli ke bagian tengah kanvas menggunakan slicing
    kanvas[y_offset:y_offset + h, x_offset:x_offset + w] = img

    # 6. Hitung parameter visual secara adaptif berdasarkan ukuran kanvas (max_dim)
    font_scale = max_dim / 1000.0
    thickness = max(1, int(max_dim / 500))
    font = cv2.FONT_HERSHEY_SIMPLEX

    # Hitung ukuran teks nama toko
    (text_w, text_h), _ = cv2.getTextSize(nama_toko, font, font_scale, thickness)

    # Buat badge/background banner untuk logo toko di pojok kiri atas
    padding = int(max_dim * 0.015)
    badge_x1 = int(max_dim * 0.03)
    badge_y1 = int(max_dim * 0.03)
    badge_x2 = badge_x1 + text_w + (padding * 2)
    badge_y2 = badge_y1 + text_h + (padding * 2)

    # Gambar background badge (warna gelap/hitam)
    cv2.rectangle(kanvas, (badge_x1, badge_y1), (badge_x2, badge_y2), (30, 30, 30), -1)

    # Tuliskan nama toko di dalam badge
    text_x = badge_x1 + padding
    text_y = badge_y1 + text_h + padding
    cv2.putText(kanvas, nama_toko, (text_x, text_y), font, font_scale, (255, 255, 255), thickness)

    return kanvas

# === PENGUJIAN PADA GAMBAR LANSKAP DAN POTRAIT ===

# 1. Baca gambar asli
img_asli = cv2.imread('/content/drive/MyDrive/PCVK/729ad62c-aa04-4126-9758-7dc8d2a17e7d.jpg')
img_asli = cv2.cvtColor(img_asli, cv2.COLOR_BGR2RGB)

# 2. Buat simulasi gambar Lanskap (lebar > tinggi)
h_orig, w_orig = img_asli.shape[:2]
img_lanskap = cv2.resize(img_asli, (w_orig, int(w_orig * 0.6)))

# 3. Buat simulasi gambar Potrait (tinggi > lebar)
img_potrait = cv2.resize(img_asli, (int(h_orig * 0.6), h_orig))

# 4. Proses kedua gambar menggunakan fungsi adaptif
thumb_lanskap = buat_thumbnail_persegi(img_lanskap, "YANUAR STORE")
thumb_potrait = buat_thumbnail_persegi(img_potrait, "YANUAR STORE")

# 5. Tampilkan hasil
fig, axes = plt.subplots(1, 2, figsize=(12, 6))

axes[0].imshow(thumb_lanskap)
axes[0].set_title(f'Thumbnail Lanskap 1:1\n({thumb_lanskap.shape[1]}x{thumb_lanskap.shape[0]})')
axes[0].axis('off')

axes[1].imshow(thumb_potrait)
axes[1].set_title(f'Thumbnail Potrait 1:1\n({thumb_potrait.shape[1]}x{thumb_potrait.shape[0]})')
axes[1].axis('off')

plt.tight_layout()
plt.show()
```
